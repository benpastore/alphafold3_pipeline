```python
import os


def parse_fasta(fasta) : 
    d = {}
    seq = ''
    header = None
    with open(fasta, 'r') as f : 
        for line in f : 
            if line.startswith(">") : 
                if header : 
                    #print(header)
                    d[header] = seq
                    header = line.strip().replace(">", "").split("|")[1]
                    seq = ''
                else : 
                    header = line.strip().replace(">", "").split("|")[1]

            else : 
                seq += line.strip()            
        else :
            d[header] = seq
            
    f.close()
    
    return d

def subset_fasta(fasta, gene_list) : 

    total = 0
    lines = ""
    fa_dict = parse_fasta(fasta)
    genes = []
    with open(gene_list, 'r') as f : 
        for line in f : 
            info = line.strip().split("\t")
            genes.append(info[0])
    f.close()
    
    for gene in genes : 
        if gene in fa_dict.keys() : 
            lines += f">{gene}\n{fa_dict[gene]}\n"
            total += 1


    op = open("c_elegans_E3.fa", 'w')
    op.write(lines)
    op.close()

    print(total)
            
        
        
        
    
```

```python
subset_fasta("./UP000001940_6239.fasta", "./uniprot_E3_all.txt")
```
