```python
import re
```

```python
def read_fasta(fasta) : 
    
    fa_dict = {}
    header = False
    seq = ''
    with open(fasta, 'r') as f : 
        for line in f : 
            if line.startswith(">") : 
                if not header : 
                    #header = line.strip().split(" ")[0].replace(">", "").split(",")[0]
                    match = re.search(r"GN=([\w\.\-]+)", line.strip())
                    header = match.group(1)
                else :
                    fa_dict[header] = seq 
                    match = re.search(r"GN=([\w\.\-]+)", line.strip())
                    header = match.group(1)
                    seq = ''
            else : 
                seq += line.strip()
        else : 
            fa_dict[header] = seq
            
    f.close()
    return fa_dict
```

```python
def read_list(file) : 

    my_list = []
    with open(file, 'r') as f :
        for line in f :
            my_list.append(line.strip())
    f.close() 

    return my_list
```

```python
npps = read_list("../af3_inputs/npp_list.txt")
```

```python
def make_fasta(fa_dict : dict, outname : str, write_all = True, gene_list = '') : 

    lines = ''
    if write_all : 
        for gene, seq in fa_dict.items() : 
            lines += f">{gene}\n{seq}\n"
    else : 
        for gene in gene_list : 
            lines += f">{gene}\n{fa_dict[gene]}\n"

    op = open(outname, 'w')
    op.write(lines)
    op.close()
    
```
